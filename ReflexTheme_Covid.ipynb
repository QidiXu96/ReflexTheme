```python
def build_files_list(folder_path):
    pattern = re.compile(r"^Interview_(\d+)_chunk_(\d+)\.docx$")
    grouped = defaultdict(list)

    for filename in os.listdir(folder_path):
        match = pattern.fullmatch(filename)
        if match:
            transcript_id = int(match.group(1))
            chunk_id = int(match.group(2))

            full_path = os.path.join(folder_path, filename)
            grouped[transcript_id].append((chunk_id, full_path))

    files_list = []

    for transcript_id in sorted(grouped):
        chunks_sorted = sorted(grouped[transcript_id], key=lambda x: x[0])
        files_list.append([path for _, path in chunks_sorted])

    return files_list
```
